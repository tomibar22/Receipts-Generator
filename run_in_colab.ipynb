# Receipts Generator — Google Colab Guide

Automates receipt generation by connecting your **Notion** database to **Green Invoice** (חשבונית ירוקה).

### What It Does

| Step | Action |
|---|---|
| 1 | Authenticates with Green Invoice API |
| 2 | Queries your Notion database for last month's unmarked income entries |
| 3 | Creates a **type 400 receipt** in Green Invoice for each entry |
| 4 | Checks the **קבלה** checkbox in Notion so it won't run again |

### Notes
- Run **once a month** — it automatically targets the previous month
- Special case: running after Dec 22 generates receipts for the current year
- Notion entries must have **דו״ח** ✅ and **קבלה** + **חשבונית** ☐ to be picked up
- Dependency warnings from Colab are safe to ignore

---

### Required Environment Variables

| Variable | Source |
|---|---|
| `NOTION_SECRET` | Notion integration token |
| `DATABASE_ID` | Your Notion database ID |
| `MORNING_SECRET` | Green Invoice API secret |
| `MORNING_ID` | Green Invoice API user ID |
